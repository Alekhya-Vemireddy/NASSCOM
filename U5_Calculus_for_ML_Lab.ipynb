```python
import numpy as np
import sympy as sp
x, y = sp.symbols('x y')
sp.init_printing()
np.random.seed(42)
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)
```

```python
def f(x):
    return x**2
h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')
```

```python
expr = x**2
deriv = sp.diff(expr, x)
print('d/dx (x**2) =', deriv)
print('Symbolic f\'(3) =', deriv.subs(x, 3))
```

```python
def g(x):
    return x**3 + 2*x
# 1. Numerical derivative at x = 2 (use h = 1e-6)
print('Numerical g\'(2):', round(slope_at_3, 4), ' (exact = 9)')
# 2. Symbolic derivative of x**3 + 2*x
print('Symbolic g\'(x) =', deriv)
# 3. Evaluate the symbolic derivative at x = 2 and compare
print('Symbolic g\'(2) =', deriv.subs(x, 2))
```

```python
f2 = x**2 + 3*x*y + y**2
print('df/dx =', sp.diff(f2, x))
print('df/dy =', sp.diff(f2, y))
```

```python
grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)
```

```python
h2 = x**2 * y + sp.sin(y)
dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)
grad_h = [dh_dx, dh_dy]
# 1. dh/dx and dh/dy
print('dh/dx =', dh_dx)
print('dh/dy =', dh_dy)
# 2. Assemble the gradient list
print('grad h =', grad_h)
# 3. Evaluate at (x=2, y=0)  -> hint: .subs({x: 2, y: 0})
grad_h_at = [g.subs({x: 2, y: 0}) for g in grad_h]
print('grad h at (2, 0) =', grad_h_at)
```

```python
by_hand = sp.cos(x**2) * 2*x
by_sympy = sp.diff(sp.sin(x**2), x)
print('By hand :', by_hand)
print('By SymPy:', by_sympy)
print('Match?  ', sp.simplify(by_hand - by_sympy) == 0)
```

```python
expr3 = (3*x + 1)**4
print('d/dx (3x+1)^4 =', sp.diff(expr3, x))
```

```python
# y = exp(x**2 + 1);  inner = x**2 + 1, inner' = 2x, outer' = exp(inner)

# 1. By hand (as a SymPy expression)
by_hand_deriv = sp.exp(x**2 + 1) * 2*x
print('d/dx (exp(x**2 + 1)) =', by_hand_deriv)

# 2. With sp.diff
sympy_deriv = sp.diff(sp.exp(x**2 + 1), x)
print('d/dx (exp(x**2 + 1)) =', sympy_deriv)

# 3. Confirm they match
print('Match?  ', sp.simplify(by_hand_deriv - sympy_deriv) == 0)
```

```python
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1
z1 = X @ W1
h = np.maximum(0, z1)
y_hat = h @ W2
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
dy   = 2 * (y_hat - Y) / Y.size
dW2  = h.T @ dy
dh   = dy @ W2.T
dz1  = dh * (z1 > 0)
dW1  = X.T @ dz1
print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python
lr = 0.1
W1 -= lr * dW1
W2 -= lr * dW2
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```
